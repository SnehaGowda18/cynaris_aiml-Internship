```python
# W1D3: Data Loading, Cleaning & Inspection
# Author: Sneha G R

import pandas as pd
import numpy as np
import os


# -------------------------------
# 1. Create sample dataset
# -------------------------------

data = {
    "State": [
        "Karnataka",
        "Tamil Nadu",
        "Kerala",
        "Karnataka",
        "Maharashtra"
    ],
    "Population": [
        61095297,
        72147030,
        None,
        61095297,
        112374333
    ],
    "Area": [
        191791,
        130058,
        38863,
        191791,
        307713
    ]
}


df = pd.DataFrame(data)

print("Original Dataset")
print(df)


# -------------------------------
# 2. Data Inspection
# -------------------------------

print("\nDataset Shape:")
print(df.shape)

print("\nData Types:")
print(df.dtypes)

print("\nFirst 10 Rows:")
print(df.head(10))


# -------------------------------
# 3. Check Missing Values
# -------------------------------

print("\nMissing Values:")
print(df.isnull().sum())


# -------------------------------
# 4. Data Cleaning
# -------------------------------

# Fill missing numerical values with mean

df["Population"] = df["Population"].fillna(
    df["Population"].mean()
)


# Remove duplicate rows

df = df.drop_duplicates()


# Remove extra spaces from text columns

df["State"] = df["State"].str.strip()


print("\nCleaned Dataset")
print(df)


# -------------------------------
# 5. Export Clean Dataset
# -------------------------------

df.to_csv(
    "cleaned_w1d3_states.csv",
    index=False
)

df.to_parquet(
    "cleaned_w1d3_states.parquet",
    index=False
)


# -------------------------------
# 6. Compare File Sizes
# -------------------------------

csv_size = os.path.getsize(
    "cleaned_w1d3_states.csv"
)

parquet_size = os.path.getsize(
    "cleaned_w1d3_states.parquet"
)


print("\nFile Size Comparison")
print("CSV Size:", csv_size, "bytes")
print("Parquet Size:", parquet_size, "bytes")


print("\nW1D3 Task Completed Successfully!")
```
