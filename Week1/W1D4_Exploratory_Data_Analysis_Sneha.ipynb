```python
# W1D4: Exploratory Data Analysis (EDA)

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import os

# -----------------------------
# 1. Load Dataset
# -----------------------------

df = pd.read_csv("../../sample_data.csv")

print("Dataset Loaded Successfully\n")

# Display first rows
print("First 5 Rows:")
display(df.head())


# -----------------------------
# 2. Dataset Information
# -----------------------------

print("\nDataset Shape:")
print(df.shape)


print("\nDataset Information:")
df.info()


print("\nStatistical Summary:")
display(df.describe())


print("\nMissing Values:")
print(df.isnull().sum())


print("\nDuplicate Values:")
print(df.duplicated().sum())


# -----------------------------
# 3. Observations
# -----------------------------

print("""
EDA Observations:

1. Dataset contains numerical and categorical features.
2. Statistical analysis shows the distribution and range of numerical values.
3. Missing value analysis helps identify data cleaning requirements.
4. Visualization helps understand patterns and variations in data.
5. Correlation analysis identifies relationships between numerical features.
""")


# -----------------------------
# 4. Separate Columns
# -----------------------------

numeric_cols = df.select_dtypes(include=np.number).columns
categorical_cols = df.select_dtypes(include="object").columns


print("\nNumeric Columns:")
print(list(numeric_cols))

print("\nCategorical Columns:")
print(list(categorical_cols))


# -----------------------------
# 5. Distribution Plots
# -----------------------------

for col in numeric_cols:
    plt.figure(figsize=(7,4))
    sns.histplot(df[col], kde=True)
    plt.title(f"Distribution of {col}")
    plt.xlabel(col)
    plt.ylabel("Frequency")
    plt.show()


# -----------------------------
# 6. Correlation Heatmap
# -----------------------------

if len(numeric_cols) > 1:

    plt.figure(figsize=(8,5))

    sns.heatmap(
        df[numeric_cols].corr(),
        annot=True,
        cmap="coolwarm"
    )

    plt.title("Correlation Heatmap")
    plt.show()


# -----------------------------
# 7. Top 10 Category Counts
# -----------------------------

for col in categorical_cols:

    plt.figure(figsize=(8,4))

    df[col].value_counts().head(10).plot(
        kind="bar"
    )

    plt.title(f"Top 10 Categories - {col}")
    plt.xlabel(col)
    plt.ylabel("Count")
    plt.xticks(rotation=45)

    plt.show()


# -----------------------------
# 8. Save Output
# -----------------------------

os.makedirs("outputs", exist_ok=True)

df.to_csv(
    "outputs/eda_output.csv",
    index=False
)

print("\nEDA output saved successfully:")
print("outputs/eda_output.csv")


# -----------------------------
# 9. EDA Narrative
# -----------------------------

print("""
EDA Narrative:

Exploratory Data Analysis was performed to understand the dataset structure,
quality and hidden patterns. Dataset information, statistical summary and
missing value analysis were performed to identify data quality issues.

The describe() function provided important statistical information such as
mean, minimum, maximum and standard deviation of numerical features.
Distribution plots helped understand the spread of individual numerical
columns and identify possible variations or outliers.

The correlation heatmap showed relationships between numerical features and
helped identify important feature relationships. Categorical feature analysis
was performed using top category counts to understand frequently occurring
values.

Some possible issues in the dataset include missing values, duplicate records,
and inconsistent formats. These problems should be handled before applying
machine learning algorithms.

The EDA process provides a clear understanding of the dataset and helps in
future preprocessing steps such as encoding, scaling, feature selection and
model building.
""")
```
