```python
import pandas as pd
```

```python
df = pd.read_csv("ICRISAT-District Level Data.csv")
```

```python
# Filter records where Rice production is greater than 1000 thousand tons
filtered_df = df[df["RICE PRODUCTION (1000 tons)"] > 1000]

print(filtered_df.head(10))
```

```python
# Group by State Name and calculate total Rice production
grouped_df = df.groupby("State Name")["RICE PRODUCTION (1000 tons)"].sum()

print(grouped_df)
```

```python
# Create a DataFrame containing state capitals
capital_df = pd.DataFrame({
    "State Name": [
        "Karnataka",
        "Tamil Nadu",
        "Kerala",
        "Maharashtra",
        "Andhra Pradesh"
    ],
    "Capital": [
        "Bengaluru",
        "Chennai",
        "Thiruvananthapuram",
        "Mumbai",
        "Amaravati"
    ]
})
```

```python
# Merge the original DataFrame with capital information
merged_df = pd.merge(df, capital_df, on="State Name", how="left")

print(merged_df.head())
```

```python
# Create a pivot table showing total Rice production by State
pivot = pd.pivot_table(
    df,
    values="RICE PRODUCTION (1000 tons)",
    index="State Name",
    aggfunc="sum"
)

print(pivot)
```

```python
# Remove duplicate rows
clean_df = df.drop_duplicates()

# Remove rows containing missing values
clean_df = clean_df.dropna()

print(clean_df.shape)
```

```python
# Export cleaned DataFrame to CSV
clean_df.to_csv("cleaned_data.csv", index=False)
```
